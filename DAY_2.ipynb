```python
from abc import ABC, abstractmethod

class Test(ABC):
    @abstractmethod
    def m1(self):
        pass

    @abstractmethod
    def m2(self):
        pass


class MyPython(Test):
    def m1(self):
        print("m1 implementation")

    def m2(self):
        print("m2 implementation")


m = MyPython()
m.m1()
m.m2()
```

```python
from abc import ABC, abstractmethod

class MyPython(Test):
    def m1(self):
        print("m1 implementation")

    @abstractmethod
    def m2(self):
        pass


class child(MyPython):
    def m2(self):
        print("m2 implementation")

m = child()
m.m1()
m.m2()
```

```python
class Test1:
    _b=None
    __c=None
    def __init__(self):
        self.a=101
        self._b=201
        self.__c=301
    def show1(self):
        print("a=",self.a)
        print("b=",self._b)
        print("c=",self.__c)
class Test2(Test1):
    def show2(self):
       print("a=",self.a)
       print("b=",self._b)
class Test3(Test2):
    def show3(self):
       print("a=",self.a)
       print("b=",self._b)
t1=Test1()
t1.show1()
t2=Test2()
t2.show2()
t3=Test3()
t3.show3()
```

```python
data = [[[1, 2], [3, 4]], [[5, 6], [7, 8]]]

def fun(m):
    v = m[0][0]

    for row in m:
        for element in row:
            if v < element:
                v = element

    return v

print(fun(data[0]))
```
